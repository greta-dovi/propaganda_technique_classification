##### Linear scheduler with warm-up

```python
data_size = 8762
dataloader_size = round(8762 / 16, 0)
epochs = 20
training_steps = epochs * dataloader_size
warmup_steps = 0.1 * training_steps
lr = 2e-5

while_warm_x = np.linspace(0, warmup_steps)
while_warm_y = np.linspace(0, lr)
after_warm_x = np.linspace(warmup_steps, training_steps)
after_warm_y = np.linspace(lr, 0)


plt.plot(while_warm_x, while_warm_y, "b")
plt.plot(after_warm_x, after_warm_y, "b")
ax = plt.gca()
fig = plt.gcf()
fig.set_size_inches(10, 5)
ax.yaxis.set_major_formatter(mticker.FormatStrFormatter('%.2e'))
ax.ticklabel_format(axis='y')
plt.xlabel("Training steps")
plt.ylabel("Learning rate")

plt.title("Learning rate scheduler with warm-up")
```
